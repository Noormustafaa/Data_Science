```python
import seaborn as sns 
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 

#Note seaborn library has already some Dataset so let's use some of them 
# let's see how can we load 

df=sns.load_dataset("tips")
df
```

```python
df["size"].unique()
#Note Distribution plots are only use for Quantitative data 
```

```python
sns.displot(df["total_bill"])
```

```python
sns.displot(df["total_bill"],bins=1)
```

```python
plt.subplot(1,2,1)

sns.histplot(df["total_bill"],bins=22,kde=True)

plt.subplot(1,2,2)
sns.histplot(df["total_bill"],bins=22,kde=True)
```

```python
# Now What if you need to compare 2 data's ?
#Like comparing total_bills and tips ??
# So For That we will use the jointplot

sns.jointplot(x="total_bill", y="tip", data=df,kind="scatter")

```

```python
#other type is  hex
sns.jointplot(x="total_bill", y="tip", data=df,kind="hex")
```

```python
#other type is  kde
sns.jointplot(x="total_bill", y="tip", data=df,kind="kde")
```

```python
#other type is  reg
sns.jointplot(x="total_bill", y="tip", data=df,kind="reg")
```

```python
#pairplots 
# this is very much important  ! because for analyzing we us it !
```

```python
sns.pairplot(df,hue="sex")
# If the plot is dominated by the color blue, it usually means:
# 1. Class Imbalance: The category colored in blue has significantly more data points than others.
# 2. Overlapping Data: Blue data points are plotted directly over other categories, hiding them underneath.

# Step 1: Check the exact count of each category to confirm if there is an imbalance
print(df["sex"].value_counts())

# Step 2: Use transparency (alpha) to make overlapping points visible underneath blue points
import seaborn as sns

sns.pairplot(df, hue="sex", plot_kws={"alpha": 0.5})
```
