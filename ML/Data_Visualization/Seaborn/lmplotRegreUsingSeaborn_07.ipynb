```python
import seaborn as sns
import matplotlib.pyplot as plt

#lmplot is used to plot data and regression model fits across a FacetGrid. It combines regplot() and FacetGrid. It is used to visualize the relationship between two variables and can also be used to fit a linear regression model to the data.
```

```python
tips=sns.load_dataset('tips')
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips)
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips, hue='sex',)
```
