```python
import numpy as np
from sklearn.cluster import KMeans

# 1. Sample Data (Unlabeled 2D features: e.g., [Age, Income in $1k])
X = np.array([
    [18, 15],  # Person 1
    [20, 18],  # Person 2
    [22, 21],  # Person 3
    [45, 80],  # Person 4
    [49, 85],  # Person 5
    [52, 90]   # Person 6
])

# 2. Initialize K-Means with K = 2 clusters
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

# 3. Train/Fit the model on unlabeled data
kmeans.fit(X)

# 4. Get cluster assignments and cluster centers (centroids)
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Cluster assigned to each point:", labels)
print("Center coordinates of each cluster:\n", centroids)

# 5. Predict cluster for a brand new data point [Age: 21, Income: 20]
new_person = [[21, 20]]
prediction = kmeans.predict(new_person)

print(f"\nNew Person {new_person} belongs to Cluster: {prediction[0]}")
```
