```python
#A histogram is a graphical representation used to visualize the distribution of continuous numerical data. It groups individual numerical data points into continuous, non-overlapping intervals called bins (or classes) and plots the number of occurrences (frequency) in each bin.
#Key Components of a HistogramBins / Intervals ($X$-axis): Continuous numerical ranges (e.g., ages 20–30, 30–40, 40–50).Frequency ($Y$-axis): The total count of data points that fall within each bin interval.No Gaps Between Bars: Unlike bar charts, adjacent bars touch each other directly to demonstrate that the underlying data scale is continuous.
#Concrete Example: Student Exam Scores

#Suppose an exam is evaluated out of 100 points across 150 students. Rather than listing 150 individual scores, data is grouped into score ranges (bins of size 10):
#This distribution reveals a normal bell-curve shape, showing that most students scored around 70–80 points.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# 1. Continuous Numerical Data generate karen
np.random.seed(42)
scores = np.random.normal(loc=70, scale=12, size=150) # Average score 70
scores = np.clip(scores, 40, 100)                      # Range 40 to 100

df = pd.DataFrame({'Exam_Scores': scores})

# 2. Theme aur figure size set karen
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 3. Seaborn se Histogram plot karen
ax = sns.histplot(
    data=df, 
    x='Exam_Scores', 
    bins=8,               # Kitne ranges/bins mein divide karna hai
    kde=True,             # Continuous curve (Kernel Density Estimation) line overlay
    color='#2b5c8f', 
    edgecolor='black'
)

# 4. Labels aur Title add karen
plt.title('Distribution of Student Exam Scores', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Exam Scores (Continuous Range)', fontsize=12, labelpad=10)
plt.ylabel('Number of Students (Frequency)', fontsize=12, labelpad=10)

plt.tight_layout()
plt.show()
```

```python
#Code Highlights:

#bins=8: Numerical values ko 8 continuous intervals mein divide karta hai.

#kde=True: Histogram ke upar ek smooth probability density curve (Kernel Density Estimation) add karta hai jo data ke overall shape/trend ko dikhata hai.
```
