```python
import pandas as pd
import matplotlib.pyplot as plt

# DataFrame create karen
df = pd.DataFrame({
    'Category': ['SUV', 'Sedan', 'Truck', 'Electric (EV)'],
    'Sales': [45, 30, 20, 15]
})

# Direct Pandas bar plot
df.plot(x='Category', y='Sales', kind='bar', legend=False, color='steelblue', edgecolor='black')

plt.title('Monthly Vehicle Sales')
plt.ylabel('Units Sold')
plt.xticks(rotation=0) # Categories ko seedha dikhane ke liye
plt.show()
```
