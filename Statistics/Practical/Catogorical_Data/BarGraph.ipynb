```python
#A bar graph (or bar chart) is a visual representation of data that uses rectangular bars to show comparisons among distinct groups. The height or length of each bar is directly proportional to the numerical value or frequency it represents.
#Key Components of a Bar GraphAxes: The horizontal axis ($X$-axis) typically holds the group names, while the vertical axis ($Y$-axis) holds the numerical measurement scale (counts, percentages, or sums)
#Bars: Rectangles of equal width where the height indicates the category's value.

#Gaps: Spaces intentionally left between the bars to highlight that each category is independent.

#How Bar Graphs Handle Categorical Data

#Categorical data classifies information into non-numerical groups or labels (e.g., shirt sizes, preferred social media platforms, or regions).
#Bar graphs are the primary visualization tool for categorical data for three key reasons:

#Clear Visual Separation: The physical space between bars reinforces that the categories are discrete entities with no intermediate numeric values between them.

#Direct Comparison: Taller or longer bars immediately convey which category occurs most or least frequently.

#Flexible Sorting: Categories can be ordered alphabetically, logically (e.g., Small, Medium, Large), or by frequency (Pareto chart style) without altering the underlying data.




```

```python
import matplotlib.pyplot as plt 

# 1. Categorical Data define karen
categories = ['SUV', 'Sedan', 'Truck', 'Electric (EV)']
units_sold = [45, 30, 20, 15]

plt.Figure(figsize=(10,6))

plt.bar(categories, units_sold, color=['blue', 'green', 'red', 'orange']) # Note plt.bar() is used to create a bar graph

plt.title('Units Sold by Vehicle Type', fontsize=16, fontweight='bold') # Title of the graph

plt.xlabel('Vehicle Type', fontsize=14) # Label for X-axis

plt.ylabel('Units Sold', fontsize=16 , fontweight=12) # Label for Y-axis
 

# 5. Bars ke upar exact numbers show karne ke liye loop
for i, val in enumerate(units_sold):
    plt.text(i, val + 1, str(val), ha='center', va='bottom', fontsize=10)

# 6. Graph display karen

plt.tight_layout()
plt.show()
```
