```python
#A pie chart is a circular statistical graphic divided into slices to illustrate numerical proportions. The entire circle represents 100% of the dataset, while each individual slice represents a specific category's proportional contribution to the total.
#Key Components of a Pie ChartCircle (The Whole): Represents the sum total of all categorical frequencies (100% or $360^\circ$).Slices (Parts): Each wedge corresponds to a single category. The central angle and area of the wedge are directly proportional to the percentage it represents.Percentage Labels (autopct): Direct numeric percentages printed inside or beside each slice.
```

```python
#How Pie Charts Handle Categorical Data

#Pie charts are specifically designed for discrete categorical data when the primary goal is showing part-to-whole relationships.

#Relative Contribution: Unlike bar charts (which highlight absolute counts), pie charts quickly reveal how large or small a category is compared to the entire group.
#Fixed Baseline (100% Constraint): They work only when all categorical counts sum up to a meaningful 100% (e.g., total market share, budget allocation, or survey responses).

#Category Limit Rule: Pie charts are most effective with 5 to 7 categories maximum. Adding too many slices makes small wedges unreadable and hard to compare.
```

```python
import matplotlib.pyplot as plt
# 1. Categorical Data aur values
categories = ['SUV', 'Sedan', 'Hatchback', 'Truck', 'Electric (EV)']
shares = [80, 24, 20, 12, 8]
colors = ['#FF9999', '#66B3FF', '#99FF99', '#FFCC99', '#FFD700']  # Custom colors for each slice
explode = (0.1, 0, 0, 0, 0)  # Explode the first slice (SUV)


plt.figure(figsize=(8, 8))

# 4. Pie chart create plot karen
plt.pie(
    shares, 
    labels=categories, 
    autopct='%1.1f%%',       # Percentages show karne ke liye
    startangle=140,          # Graph ki rotation angle
    colors=colors, 
    explode=explode,         # Highlighted section
    shadow=True,             # 3D Shadow effect
    textprops={'fontsize': 11, 'weight': 'bold'}
)



# 5. Title add karen
plt.title('Vehicle Category Market Share (%)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
#Key Code Parameters:

#autopct='%1.1f%%': Automatically calculates and displays percentage values up to one decimal place inside each slice.

#explode=(0.05, 0, ...): Pulls out a specific slice (in this case, SUV) for emphasis.

#startangle=140: Rotates the chart start line to improve slice visibility and alignment.
```
