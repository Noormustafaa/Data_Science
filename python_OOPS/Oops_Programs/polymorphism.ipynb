```python
#polymorphism :Operator overloading
#When the same Operator is allowed to have diffrent meanining according to the context 

# means + operator is used to add two numbers and also to concatenate two strings
print(1 + 2) #3
print("apna" + "college") #concatenate
print([1, 2, 3] + [4, 5, 6]) # merge two lists 
# so here one operator is used to perform different operations according to the context. This is called polymorphism.

# So These All are emplicit polymorphism because we are not defining any function or class to perform this operation.

#Polymorphism (Poly = Multiple, Morph = Forms) ka simple matlab hai: "Ek hi cheez ke alag alag roop (forms)."#
```

```python
#OOPs mein polymorphism mostly Method Overriding ke zariye hoti hai. Iska matlab hai ke alag alag classes mein same name ka method hoga, lekin unka kaam alag hoga.
# Parent Class (ya General Class)
class Animal:
    def speak(self):
        print("Animal makes a sound")

# Child Class 1
class Dog(Animal):
    def speak(self):
        print("Dog barks: Woof Woof!")

# Child Class 2
class Cat(Animal):
    def speak(self):
        print("Cat meows: Meow Meow!")

# --- Polymorphism in Action ---

# Objects banana
my_dog = Dog()
my_cat = Cat()

# Loop ke zariye ek hi method 'speak()' call karna
animals = [my_dog, my_cat]

for animal in animals:
    animal.speak()  # Har animal apne hisab se sound nikalega!


  #  Same Interface / Method Name: Humne sirf speak() call kiya.

#Different Behavior: Dog ne apna barking output diya aur Cat ne meow output diya.

#Fayda: Code clean rehta hai aur naye classes add karna (jaise Cow jiska apna speak() ho) bohot asaan ho jata hai bina purana code chede.
```
