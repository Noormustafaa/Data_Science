```python
#super method: is used to call the parent class method from the child class. It is used to access the methods of the parent class from the child class.

class Car:
    def __init__(self, type):
        self.type = type

    @staticmethod
    def start():
        print("car started..")

    @staticmethod
    def stop():
        print("car stopped.")


class ToyotaCar(Car):
    def __init__(self, name, type):
        self.name = name
       # self.type = type  #Agr hum yai likhy gai to yai to ToyotaCar class ka current object hoga na k Car class ka, is liye hum super method ka use karain gai.
        super().__init__(type)  #super method ka use kar k hum parent class ka constructor call karain gai, aur type attribute ko parent class ka object banain gai.
car1 = ToyotaCar("prius", "electric")
print(car1.type)
```
