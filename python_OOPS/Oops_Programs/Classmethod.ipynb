```python
#Class method 
#A Class method is a method that is bound to the class and not the instance of the class. It can be called on the class itself, or on an instance of the class. Class methods take a reference to the class as their first parameter, which is usually named cls.

# Note: Static methods are common for complete class and for all objects we create one specific method , and all object can use it . 
# because static methods does not change the instance attributes or class attributes, they are used to perform a specific task that is related to the class but does not require access to any instance or class attributes.

#Note: static method can't access or modify class state  or instance state. They are used to perform a specific task that is related to the class but does not require access to any instance or class attributes.
```

```python
class Person:
    name = "Noor"

    # self.name = name          # Updates the instance attribute
    # Person.name = name        # Updates the class attribute using the class name directly

    def change(self, name):
        self.__class__.name = name  # Updates the class attribute dynamically using self.__class__


p1 = Person()
p1.change("Rajpar")
print(p1.name)  # Output: Rajpar
```

```python
# So class methods are used to access or modify class attributes, and they can be called on the class itself or on an instance of the class. They are defined using the @classmethod decorator and take a reference to the class as their first parameter, which is usually named cls.
# so For Writing this we use class decorator @classmethod and then we write def method_name(cls, other_parameters): and then we can access class attributes using cls.attribute_name.
class Person:
    name = "anonymous"

    # def changeName(obj, name):
    #     self.__class__.name = "Rahul"

    @classmethod
    def changeName(cls, name):
        cls.name = name


p1 = Person()
p1.changeName("rahul kumar")
print(p1.name)
print(Person.name)
```
